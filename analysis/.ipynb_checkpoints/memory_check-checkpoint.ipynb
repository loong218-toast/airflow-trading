```python
import psutil

# 1. Total Laptop Memory
vm = psutil.virtual_memory()
total_gb = vm.total / (1024**3)
used_gb = vm.used / (1024**3)
available_gb = vm.available / (1024**3)

# 2. Specific Jupyter Usage
process = psutil.Process()
jupyter_gb = process.memory_info().rss / (1024**3)

print(f"=== SERVER LAPTOP RAM STATUS ===")
print(f"Total Physical RAM:  {total_gb:.2f} GB")
print(f"Total Used (All):    {used_gb:.2f} GB ({vm.percent}%)")
print(f"Total Available:     {available_gb:.2f} GB")
```
